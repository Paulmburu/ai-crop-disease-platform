# 03 - Baseline model
Trains MobileNetV2 (frozen backbone) and evaluates on the test split.
Tip: `num_workers=0` is the safest choice inside Jupyter on macOS.

```python
%matplotlib inline
import os

os.chdir('..')  # run everything from ml-training/ so data/ and outputs/ resolve
```

```python
from crop_ml.config import TrainConfig
from crop_ml.train import fit

cfg = TrainConfig(epochs=5, num_workers=0)
history = fit(cfg)
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
for k in ('train_loss', 'val_loss'): ax[0].plot(history[k], label=k)
for k in ('train_acc', 'val_acc'): ax[1].plot(history[k], label=k)
ax[0].legend(); ax[1].legend()
```

```python
from crop_ml.evaluate import evaluate

summary = evaluate(cfg.checkpoint_path, cfg.output_dir / 'splits.json', cfg.output_dir, num_workers=0)
```
