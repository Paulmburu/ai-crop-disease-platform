# 01 - Gradient descent from scratch
The code lives in `crop_ml.theory`; this notebook runs it and shows the plots.

```python
%matplotlib inline
import os

os.chdir('..')  # run everything from ml-training/ so data/ and outputs/ resolve
```

```python
from crop_ml.theory.gradient_descent import generate_data, plot_results, train

X, y = generate_data()
w, b, losses = train(X, y, lr=0.05, epochs=200)
print(f'w={w:.3f} b={b:.3f}  (true 3, 2)')
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
ax[0].scatter(X, y, alpha=.5); ax[0].plot(sorted(X), [w*x+b for x in sorted(X)], 'r')
ax[1].plot(losses); ax[1].set_yscale('log'); ax[1].set_title('loss')
plot_results(X, y, w, b, losses)
```

```python
from crop_ml.theory.lr_experiment import run_all

for label, h in run_all().items():
    plt.plot([min(v, 1e3) for v in h], label=label)
plt.yscale('log'); plt.legend(); plt.title('learning rate vs convergence')
```
